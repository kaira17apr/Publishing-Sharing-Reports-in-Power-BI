**Q1. Explain why publishing and sharing reports is considered more important than just building dashboards in Power BI Desktop**

**Ans-**

Publishing and sharing reports in Power BI is more important than just building dashboards in Power BI Desktop because:

*   **Collaboration** → Reports can be shared across teams, ensuring everyone works with the same insights.
*  **Accessibility** → Shared reports are available on the cloud, mobile, and web, not limited to your local PC.
*   **Real-time updates** → Published reports auto-refresh with the latest data, keeping decisions accurate.
*   **Scalability** → Reports can reach large audiences, unlike dashboards stuck in Desktop.
*  **Decision-making** → Shared insights drive collective, fact-based actions across the organization.



**Q2. Define Power BI Service and explain its role in comparison to Power BI Desktop in the reporting lifecycle.**

**Ans-**

Power BI Service is the cloud platform for publishing, sharing, and consuming reports, while Power BI Desktop is the local tool for building and modeling them. Together, they form the full reporting lifecycle: Desktop is for creation, Service is for collaboration and distribution.

**Role in Reporting Lifecycle vs Desktop**

| **Stage** | **[Power BI Desktop](ca://s?q=Power_BI_Desktop_definition)** | **[Power BI Service](ca://s?q=Power_BI_Service_definition)** |
| --- | --- | --- |
| **Purpose** | Build reports, data models, and visuals | Publish, share, and consume reports |
| **Environment** | Windows desktop app | Cloud-based, browser/mobile |
| **Data Modeling** | Full modeling (relationships, DAX, Power Query) | Limited modeling, mainly consumption |
| **Report Creation** | Rich design canvas with all features | Web editing, dashboards, apps |
| **Collaboration** | Single-user, file-based (.pbix) | Multi-user, workspaces, apps, comments |
| **Refresh** | Manual refresh only | Automated scheduled refresh |
| **Distribution** | Save/share files manually | Controlled sharing with roles & security |
| **Cost** | Free | Requires license for sharing |

**Q5. Differentiate between Report and Dashboard in Power BI based on structure and usage.**

**Ans-**

In Power BI, a Report is a multi-page, interactive view built on a single dataset, while a Dashboard is a single-page summary in the Power BI Service created by pinning visuals from one or more reports. Reports are for deep analysis; dashboards are for quick monitoring.

**Key Differences Between Report and Dashboard**

| **Aspect** | **[Report](ca://s?q=Power_BI_Report_definition)** | **[Dashboard](ca://s?q=Power_BI_Dashboard_definition)** |
| --- | --- | --- |
| **Structure** | Multi-page, interactive visuals from one dataset | Single-page canvas with pinned tiles from multiple reports/datasets |
| **Creation Tool** | Built in **Power BI Desktop** and published to Service | Created only in **Power BI Service** |
| **Interactivity** | Full features: slicers, filters, drill-through, cross-highlighting | Limited: no slicers, minimal interactivity; tiles link back to reports |
| **Usage Purpose** | Detailed analysis, storytelling, exploring trends | At-a-glance monitoring of KPIs across areas |
| **Data Source** | Single semantic model/dataset | Can combine visuals from multiple datasets/reports |
| **Audience** | Analysts and users needing deep insights | Executives/managers needing quick overview |
| **Refresh** | Manual in Desktop, scheduled in Service | Auto-refresh from underlying reports/datasets |

**Q6. Explain the difference between the following three sharing methods in Power BI Service:**

*   **Direct Report Sharing**
*   **Workspace Access**
*   **Power BI Apps**
**Also mention one use case for each.**


**Ans-**

 **Sharing Methods in Power BI Service**

1.   **Direct Report Sharing**

*   **Meaning:** Share a specific report directly with individuals by email or link.
*  **Usage:** Quick, one-to-one or small group sharing.
*  **Use Case:** A data analyst shares a monthly sales report directly with the sales manager.

2.   **Workspace Access**

*   **Meaning:** Give users access to a workspace where multiple reports, datasets, and dashboards are stored.
*   **Usage:** Collaborative environment for teams to co-develop and manage reports.
*   **Use Case:** A project team working on customer analytics collaborates in a shared workspace to edit and update reports together.

3.   **Power BI Apps**

*   **Meaning:** Bundle multiple dashboards and reports into a single app for distribution.
*   **Usage:** Best for large-scale, structured sharing with many users.
*   **Use Case:** An organization publishes a Finance App containing dashboards for revenue, expenses, and forecasts for all executives.


**Q7. Describe the four workspace roles available in Power BI and explain how access control is managed using these roles.**

**Workspace Roles in Power BI**



1.   **Viewer**

*   **Access:** Can only view reports, dashboards, and datasets.
*   **Control:** No editing or sharing rights.
*   **Use Case:** Executives who just need to monitor KPIs without making changes.

2.   **Contributor**

*   **Access:** Can create, edit, and publish reports/dashboards within the workspace.
*  ** Control:** Cannot manage workspace settings or permissions.
*   **Use Case:** Analysts who design reports but don’t manage overall workspace.

3.   **Member**

*   **Access:** Can edit, publish, and share content; also manage datasets and refresh schedules.
*   **Control:** Moderate control but cannot assign roles or delete workspace.
*   **Use Case:** Team leads who manage report updates and share them with stakeholders.

4.   **Admin**

*   **Access:** Full control — manage roles, permissions, settings, and delete workspace if needed.
*   **Control:** Highest authority in workspace.
*   **Use Case:** IT or BI managers who oversee governance and security.

**Q8. Why are Power BI Apps considered more secure and scalable than direct sharing?**
**Explain this from a governance and enterprise usage perspective.**

**Ans-**

Power BI Apps are considered more secure and scalable than direct sharing because they provide centralized governance, controlled distribution, and enterprise-level management of reports and dashboards.

**Why Apps Are More Secure**

*   **Centralized governance** → Admins control access at the app level, reducing risks of accidental oversharing.
*   **Role-based security** → Permissions are managed consistently across all reports in the app.
*   **Data protection** → Sensitive datasets are shielded by workspace rules, not exposed through ad-hoc links.


**Why Apps Are More Scalable**

*   **Enterprise distribution** → Apps bundle multiple reports/dashboards into one package for large audiences.
*  **Ease of updates** → Changes in the workspace automatically flow to the app without re-sharing links.
*   **Consistency** → Everyone sees the same version of reports, avoiding duplication or outdated files.

**Governance & Enterprise Perspective**

*   **Direct Sharing** → Good for small, quick access but hard to track at scale.
*   **Apps** → Provide structured, governed access with audit trails, compliance, and controlled rollout across departments.



**Q9. Explain Row-Level Security (RLS) in Power BI.**

**How does RLS ensure that different users see different data using the same report?**

Row-Level Security (RLS) in Power BI restricts data access at the row level so each user only sees the data they are authorized to view. It ensures different users see different slices of the same report by applying filters based on their identity or role.

**What is RLS?**

*   Row-Level Security is a feature that limits data visibility for users.
*   It uses roles and DAX filters defined in the data model.
*   When a report is published, these rules automatically apply whenever a user opens the report.


**How RLS Works**

1.   Define roles → In Power BI Desktop, create roles (e.g., “India Sales Manager”).
2.   Apply filters → Use DAX expressions to restrict rows (e.g., [Country] = "India").
3.   Assign users → In Power BI Service, map users/groups to roles.
4.   Dynamic RLS → Use functions like USERPRINCIPALNAME() so filters apply automatically based on login credentials.


**How Different Users See Different Data**

*   Example: A single Sales Report is published.
    *   Manager A (India) → Sees only rows where Country = India.
    *  Manager B (Dubai) → Sees only rows where Country = Dubai.
*   Both use the same report, but RLS ensures personalized views without duplicating reports.


**Q10.Assume an organization shares reports without using Apps, workspace roles, or RLS.**

**Discuss any three risks or issues that may arise related to security, governance, or data misuse, and explain how Power BI best practices help prevent them.**

**Ans-**

If an organization shares reports without using Apps, workspace roles, or Row-Level Security (RLS), several risks can arise:

**Risks of Not Using Best Practices**


1.   **Data leakage**

*   Reports may be overshared via direct links, exposing sensitive information to unauthorized users.
*   Example: A financial report link accidentally forwarded outside the company.


2.   **Lack of governance**

*   No centralized control over who has access, making it hard to track usage or enforce compliance.
*   Example: Multiple versions of the same report circulate, leading to confusion and audit issues.

3.  **Data misuse**

*   Without RLS, all users see full datasets, which can lead to misuse of confidential or restricted data.
*   Example: Regional managers viewing competitor regions’ sales data they shouldn’t access.
